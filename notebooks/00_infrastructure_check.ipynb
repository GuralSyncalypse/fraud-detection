# PHASE 1 - Infrastructure check
Notebook này xác minh Jupyter kết nối Spark standalone cluster và Spark đọc/ghi trực tiếp HDFS.

```python
from pyspark.sql import SparkSession

spark = (SparkSession.builder
         .appName("phase-1-notebook-check")
         .getOrCreate())
print("Spark version:", spark.version)
print("Spark master:", spark.sparkContext.master)
print("HDFS defaultFS:", spark.sparkContext._jsc.hadoopConfiguration().get("fs.defaultFS"))
```

```python
df = spark.range(1000).repartition(4)
assert df.count() == 1000
path = "hdfs:///financial/_health/notebook-check"
df.write.mode("overwrite").parquet(path)
assert spark.read.parquet(path).count() == 1000
print("OK - Spark cluster và HDFS giao tiếp thành công:", path)
```
