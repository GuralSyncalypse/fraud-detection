# 05 - Random Forest
Load các model None, Class Weight và Undersampling. Feature importance thể hiện mức đóng góp dự báo của model, không chứng minh quan hệ nhân quả.

```python
import json
from pyspark.ml.classification import RandomForestClassificationModel
rf = RandomForestClassificationModel.load('hdfs:///financial/models/random_forest/class_weight')
with open('/workspace/output/results/phase3_summary.json', encoding='utf-8') as stream:
    feature_names = json.load(stream)['features']['names']
top_features = sorted(zip(feature_names, rf.featureImportances.toArray()), key=lambda item: item[1], reverse=True)[:20]
top_features
```
