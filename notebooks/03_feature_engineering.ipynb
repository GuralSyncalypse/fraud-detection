# 03 - Feature Engineering và Train/Test Split
Preprocessor được fit trên train, sau khi chia dữ liệu bằng hash cố định. `class`, `transaction_id`, `client_id`, `card_id` và `merchant_id` không nằm trong feature vector.

```python
import json
from pyspark.ml import PipelineModel
train = spark.read.parquet('hdfs:///financial/processed/ml_splits/train')
test = spark.read.parquet('hdfs:///financial/processed/ml_splits/test')
preprocessor = PipelineModel.load('hdfs:///financial/models/preprocessor')
train.groupBy('class').count().show()
test.groupBy('class').count().show()
```

```python
sample = preprocessor.transform(train.limit(5))
sample.select('transaction_id', 'class', 'features').show(truncate=False)
with open('/workspace/output/results/phase3_summary.json', encoding='utf-8') as stream:
    phase3 = json.load(stream)
phase3['features']
```
