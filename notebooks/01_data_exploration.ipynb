Dữ liệu giao dịch và nhãn được đọc trực tiếp từ HDFS. Không dùng `toPandas()` trên dữ liệu đầy đủ.

```python
import sys
sys.path.insert(0, '/workspace')
from src.preprocessing import create_spark_session, read_raw_transactions, read_normalized_labels, collect_raw_profile
spark = create_spark_session('01-data-exploration')
spark.sparkContext.setLogLevel('WARN')
```

```python
transactions = read_raw_transactions(spark, 'hdfs:///financial/raw/transactions_data.csv')
labels = read_normalized_labels(spark, 'hdfs:///financial/raw/fraud_labels.csv')
transactions.printSchema()
transactions.show(5, truncate=False)
```

```python
profile = collect_raw_profile(transactions)
profile
```

```python
from pyspark.sql import functions as F
label_total = labels.count()
(labels.groupBy('label').count()
 .withColumn('percentage', F.round(F.col('count') / F.lit(label_total) * 100, 6))
 .orderBy('label')
 .show())
```
