## Giao dịch cần xem xét

Top transactions được xếp theo fraud probability của RF undersampling. `actual_class` chỉ tồn tại vì đây là Test Set phục vụ đánh giá.

```python
suspicious = spark.read.parquet('hdfs:///financial/predictions/suspicious_transactions')
suspicious.orderBy('fraud_probability', ascending=False).show(20, truncate=False)
```
